```python
filename = dbutils.widgets.get("file_name")
tableschema = dbutils.widgets.get("table_schema")
tablename = dbutils.widgets.get("table_name")

spark.sql(f"create schema if not exists {tableschema}")

df = spark.read.json(f"abfss://{container}@{storage_account}.dfs.core.windows.net/bronze/{filename}")
df.write.mode("overwrite").saveAsTable(f"{tableschema}.{tablename}")
```
